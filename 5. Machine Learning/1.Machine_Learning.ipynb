# **🤖 Machine Learning**
---

<br/>

### **📝 Definition**
---

- **Machine Learning** is the science of getting computers to learn and act like humans do, impove their learning over
time in autonomous fashion, by feeding them data and information in the form of observations and real-world interactions.

<br/>

### **🤖 How does Machine Learning Work ?**
---

- A Machine Learning system learns from historical data, builds the prediction models, and whenever it receives new data,
predicts the output for it.

### **Classifications of Machine Learning**
---

<br/>

#### **✅ Supervised Learning:**

- **Supervised learning** is a type of machine learning in which a model is trained using **labeled data**. This means each training example includes both:

<ul style="margin-left: 50px;">
    <li> <b>Input (features):</b> the data given to the model.</li>
    <li> <b/>Output (label/target):</b> the correct answer the model should learn to predict.</li>
</ul>

- The goal is for the model to learn the relationship between the inputs and outputs so it can accurately predict the output for new, unseen data.

<br/>

#### **🔺 Types of Supervised Learning:**
<ol style="margin-left: 50px; ">
    <li> &ensp; <b>Classification:</b> &ensp; Predicts categories or classes.</li>
    <li> &ensp; <b>Regression</b> &ensp; Predicts continuous numerical values.</li>
</ol>

<br/>

#### ✅ Unsupervised Learning.

- **Unsupervised Learning** is a type of Machine Learning where the model learns from **unlabeled data**.

- Unlike supervised learning, there is no predefined target/output column. The algorithm tries to discover hidden patterns, structures, groups, or relationships in the data.

<br/>

#### **🔺 Types of Unsupervised Learning:**
<ol style="margin-left: 50px;">
    <li> &ensp; <b>Clustering:</b> &ensp; Groups similar data points together.</li>
    <li> &ensp; <b>Dimensionality Reduction:</b> &ensp; Reduces the number of features while trying to preserve important information.</li>
    <li> &ensp; <b>Association Rule Learning:</b> &ensp; Finds relationships between items or events.</li>
</ol>

<br/>

#### ✅ Reinforcement Learning.

- **Reinforcement Learning** is a type of Machine Learning where an **agent learns by interacting with an environment** and receiving **rewards or penalties**.

- Unlike supervised learning, there is no dataset containing the correct answer for every situation. The agent learns which actions produce better long-term results.

<br/>

#### **🔺 Key Components of Reinforcement Learning:**
<ol style="margin-left: 50px;">
    <li> &ensp; <b>Agent:</b> &ensp; The learner/decision maker.</li>
    <li> &ensp; <b>Environment:</b> &ensp; The world in which the agent operates.</li>
    <li> &ensp; <b>State:</b> &ensp; The current situation of the environment.</li>
    <li> &ensp; <b>Action:</b> &ensp; What the agent can do.</li>
    <li> &ensp; <b>Reward:</b> &ensp; Feedback received after taking an action.</li>
    <li> &ensp; <b>Policy:</b> &ensp; The strategy used by the agent to decide what action to take.</li>
</ol>
