## **_Practical Example:_**
---

<br/>

### **_Scenerio:_**
- **Imagine you are work for an e-commerce company, and your team is responsible for analyzing customer purchase data. You want to determine whether a new website design has led to a significant increase in the average purchase amount compared to the old design.**

<br/>

### **_Data:_**
- **You have collected data from a random sample of 30 customers who made purchases on the old website design and 30 customers who made purchases on the new website design. You have the sample means, sample standard deviations, and sample sized for both groups.**
<br/>

- **Old Design Data** = [45.2, 42.8, 38.9, 43.5, 41.0, 44.6, 40.5, 42.7, 39.8, 41.4, 44.3, 39.7, 42.1, 40.6, 43.0, 42.2, 41.5, 39.6, 44.0, 43.1, 38.7, 43.9, 42.0, 41.9, 42.8, 43.7, 41.3, 40.9, 42.5, 41.6]

<br/>

- **New Design Data** = [48.5, 49.1, 50.2, 47.8, 48.7, 49.9, 48.0, 50.5, 49.8, 49.6, 48.2, 49.9, 50.3, 49.4, 50.1, 48.6, 48.3, 49.0, 50.0, 48.4, 49.3, 49.5, 48.8, 50.6, 50.4, 48.1, 49.2, 50.7, 50.8]

<br/>

- **Population Std** = 2.5

#### **Importing Neccessary Libraries**

```python
import numpy as np
import scipy.stats as st
```

#### **Old Data & New Data**

```python
old_design = np.array([45.2, 42.8, 38.9, 43.5, 41.0, 44.6, 40.5, 42.7, 39.8, 41.4, 44.3, 39.7, 42.1, 40.6, 43.0, 42.2, 41.5, 39.6, 44.0, 43.1, 38.7, 43.9, 42.0, 41.9, 42.8, 43.7, 41.3, 40.9, 42.5, 41.6])

new_design = np.array([48.5, 49.1, 50.2, 47.8, 48.7, 49.9, 48.0, 50.5, 49.8, 49.6, 48.2, 49.9, 50.3, 49.4, 50.1, 48.6, 48.3, 49.0, 50.0, 48.4, 49.3, 49.5, 48.8, 50.6, 50.4, 48.1, 49.2, 50.7, 50.8])
```

#### **Given Values for (Z-Test)**

```python
population_standard_deviation = 2.5
num_samples = len(new_design)
alpha_value = 0.05
mean_old_design = np.mean(old_design)
mean_new_design = np.mean(new_design)
```

#### **Z-Calculated**

```python
z_calculated = (mean_new_design - mean_old_design) / (population_standard_deviation/np.sqrt(num_samples))

print(z_calculated)
```

```python
z_table = st.norm.ppf(1-alpha_value)

print(z_table)
```

```python
if z_calculated > z_table:
    print("Hₐ is Right..\nReject H₀..")
else:
    print("H₀ is Right..\nDon't Reject H₀")
```
