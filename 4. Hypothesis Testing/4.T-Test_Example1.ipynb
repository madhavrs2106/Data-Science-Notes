## **_Practical Example_**
---

- **A manufacturer claims that the average weight of bag of potato chips is 150 grams. A sample of 25 bags is taken, and average weight is found to be 148 grams, with the Standard deviation of 5 grams, Test the manufacturer's claim using one-tailed T-test with the significance level of 0.05.**

#### **Importing Neccessary Libraries**

```python
import numpy as np
import scipy.stats as st
```

#### **Given values (for T-Test)**

```python
sample_size = 25
population_mean = 150
sample_mean = 148
sample_standard_deviation = 5
alpha_value = 0.05
degrees_of_freedom = sample_size - 1
```

#### **T-Critical**

```python
t_critical = st.t.ppf(alpha_value,degrees_of_freedom)

print(t_critical)
```

#### **T-Calculated**

```python
t_calculated = (sample_mean - population_mean) / (sample_standard_deviation/np.sqrt(sample_size))

print(t_calculated)
```

#### **Conclusion**

```python
if t_calculated < t_critical:
    print("Hₐ is Right\nHence, Reject H₀")
else:
    print("H₀ is Right.")
```
