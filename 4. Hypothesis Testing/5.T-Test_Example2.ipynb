## **_Practical Example_**
---

- **A Company want to test wheather there is difference in productivity between two teams. They randomly select 20 employees from each team and record their productivity score. The mean productivity score for Team A is 80 with a standard deviation of 5, while mean productivity score for Team B is 75 with standard deviation of 6, Test at a 5% level of significance where there is a difference productivity between the two Teams.**

#### **Importing Neccessary Libraries**

```python
import numpy as np
import scipy.stats as st
```

#### **Given Values (for T-Test)**

```python
sample_size_TeamA, sample_size_TeamB = 20, 20
sample_Mean_TeamA, sample_Mean_TeamB = 80, 75
sample_STD_TeamA, sample_STD_TeamB = 5, 6
alpha_value = 0.05
degrees_of_freedom = sample_size_TeamA + sample_size_TeamB - 2
```

#### **T-Critical**

```python
t_critical = st.t.ppf(1-0.025, degrees_of_freedom)

print(t_critical)
```

#### **T-Calculated**

```python
standard_error = np.sqrt( (np.square(sample_STD_TeamA)/sample_size_TeamA) + (np.square(sample_STD_TeamB)/sample_size_TeamB) )
t_calculated = (sample_Mean_TeamA - sample_Mean_TeamB) / standard_error

print(t_calculated)
```

#### **Conclusion**

```python
if (t_calculated < t_critical) or (t_calculated > t_critical):
    print("Hₐ is Right\nHence, Reject H₀")
else:
    print("H₀ is Right")
```
