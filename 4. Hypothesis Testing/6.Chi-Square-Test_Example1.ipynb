# **Chi-Square-Test**
---

### **Identify When to use Chi-Square-Test**

- When you want to see if your **Categorical (count or frequency)** data matches expected values
- It is used for two main goals:

<ol style="margin-left: 50px;">
<li>Checking if two categorical variables are related (Test of Independence).</li>
<li>Testing if sample data fits a theoretical distribution (Goodness of Fit).</li>
</ol>

# **Practical Example**
---

- **A fair die is rolled 120 times and following results are obtained:**

<ul style="margin-left: 50px;">
<li> <b>Face 1:</b> &emsp; 22 times</li>
<li> <b>Face 2:</b> &emsp; 17 times</li>
<li> <b>Face 3:</b> &emsp; 20 times</li>
<li> <b>Face 4:</b> &emsp; 26 times</li>
<li> <b>Face 5:</b> &emsp; 22 times</li>
<li> <b>Face 6:</b> &emsp; 13 times</li>
</ul>

- **Test at a 5% level of significance whether the dice is fair.**

#### **Importing Neccessary Libraries**

```python
import numpy as np
import scipy.stats as st
```

#### **Given values(for Chi-Square-Test)**

```python
observed = np.array([22,17,20,26,22,13])
expected = np.full(6, observed.sum()/6)
alpha = 0.05
degree_of_freedom = len(observed) - 1

print(
    f" -->  Observed :  {observed}\n"
    f" -->  Expected :  {expected}"
)
```

#### **Chi-square-Critical**

```python
chi_square_critical = st.chi2.ppf(1-alpha, degree_of_freedom)

print(
    f" -->  Alpha :  {alpha}\n"
    f" -->  Degree of freedom :  {degree_of_freedom}\n\n"
    f" -->  Chi-square Critical :  {chi_square_critical}"
)
```

#### **Chi-square-Calculated**

```python
chi_square_calculated = np.sum(np.power(observed - expected, 2)/expected)

print(f" --> Chi-Square Calculated :  {chi_square_calculated}")
```

#### **Conclusion**

```python
if chi_square_calculated > chi_square_critical:
    print("Hₐ is Right\nHence, Reject H₀")

else:
    print("H₀ is Right\nDon\'t Reject H₀")
```
