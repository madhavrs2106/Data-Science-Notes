## **_Practical Example:_**
---

- **A teacher claims that the mean score of students in his class is greater than 82 with a standard deviation of 20. If a sample of 81 students was selected with the mean score of 90.**


#### **Importing Neccessary Libraries**

```python
import numpy as np
import scipy.stats as st
```

#### **Z-value**

```python
z_value = st.norm.ppf(0.95)

print(z_value)
```

#### **Given values (for Z-Test)**

```python
sample_size = 81
sample_mean = 90
hypothesized_population_mean = 82
population_standard_deviation = 20
alpha_value = 0.05
```

#### **Z Calculated**

```python
z_calculated = (sample_mean - hypothesized_population_mean) / (population_standard_deviation / np.sqrt(sample_size))

print(z_calculated)
```

#### **Z Table**

```python
z_table = st.norm.ppf(1 - alpha_value)

print(z_table)
```

#### **Conclusion**

```python
if z_table < z_calculated:
    print("Hₐ is Right..\nHence, Reject H₀..")

else:
    print("H₀ is Right..")
```
