# **Hypothesis Testing**
---

- It is a part of statistical analysis, where we test the assumptions made regarding a population parameter. It is generally used when we were to compare a single group with an external standard and two or more groups with each other.

<br/>

# **Terminology Used in Hypothesis Testing**
---
<br/>

#### **Null Hypothesis**
- Null Hypothesis is a statistical theory that suggests there is no statistical significance exists between the populations.
- It is denoted by H₀ and read as H-naught.

<br/>

#### **Alternative Hypothesis**
- An Alternative Hypothesis suggests there is a signigicant difference between the population parameters. It could be greater or smaller. Basically, it is the contrast of the Null Hypothesis.
- It is denoted by Hₐ or H₁.

<br/>

# **Some Common Test**
---

- Z - Test.
- P - Test.
- Chi - Square Test.

<br/>

# **Steps of Hypothesis Testing**
---

#### **1. State null (H₀) and alternative (H₁)**
- H₀ must always contain equality (=).
- H₁ always contains difference (≠, ≻, ≺)

<br/>

#### **2. Choose level of Significance (α)**
- It is a fixed probability of wrongly rejecting a True Null Hypothesis.
- Denoted by α and read as alpha.

#### **3. Find test Statistic**
- **Z-Test:** used when average is given, **mean & standard-deviation** of population and **size of sample data is greater than equal to 30**.
- **T-Test:** used when average is given, **mean** of population, **standard-deviation** of sample-data and **size of sample data is smaller than 30**.

<br/>

#### **4. Find Critical values**
- Z - value.
- T - value.

<br/>

#### **5. Draw your Conclusion**
- H₀ is right.
- H₁ is right.
