# **Central Limit Theorem (CLT)**
---

- The Central Limit Theorem (CLT) states that when plotting a sample distribution of means, the means of the sample will be equal to the population mean and the sample distribution will approach normal distribution with variance equal to standard error.


<br/>

# **Importing Neccesaary Libraries**
---

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

<br/>

# **Generating Random Data _(Using List Comprehension)_**
---

```python
pop_data = [np.random.randint(10,100) for i in range(10000)]

# --- Creating a dataframe (population-dataframe)
pop_dataframe = pd.DataFrame({ "Pop Data": pop_data })

pop_dataframe.head()
```

### **Non-Normal Distiribution**

```python
# --- Setting Figure size of Chart
plt.figure(figsize = (8,5))

# --- Creating KDE-line chart
sns.kdeplot(x="Pop Data", data=pop_dataframe)

# --- Adding Mean-line
plt.axvline(np.mean(pop_dataframe["Pop Data"]), color="red", linestyle="--", label=f"Mean:  {np.mean(pop_dataframe['Pop Data']):.1f}")

# --- Adding Title & Labels
plt.title("Non-Normal Distribution", fontsize=20, fontweight="bold")
plt.xlabel("Population", fontsize=14, fontweight="bold")
plt.ylabel("Density", fontsize=14, fontweight="bold")

# --- Adding Grid-lines & Chart-key
plt.grid(True, linestyle='--', linewidth=0.8)
plt.legend()

# --- Shows Chart
plt.show()
```

### **Central Limit Theorem**
- Converting **Non-Normal Distribution** to **Normal Distribution**.

```python
# --- Creating a list (sample-mean) of containing  means of sample-data
sample_mean = []

for no_sample in range(50):
    # --- Creating a list (sample-data)
    sample_data = []

    for data in range(500):
        # --- Adds random data from population-dataframe to sample-data
        sample_data.append(np.random.choice(pop_data))

    # --- Adds mean of sample-data to sample-mean
    sample_mean.append(np.mean(sample_data))
```

```python
sampleMean_df = pd.DataFrame({ "Mean": sample_mean })

sampleMean_df.head()
```

### **Normal Distribution (CLT)**

```python
# --- Setting Figure size of Chart
plt.figure(figsize = (8,5))

# --- Creating KDE-line chart
sns.kdeplot(x="Mean", data=sampleMean_df)

# --- Adding Mean-line
plt.axvline(np.mean(sampleMean_df["Mean"]), color="red", linestyle="--", label=f"Mean:  {np.mean(sampleMean_df['Mean']):.1f}")

# --- Adding Title & Labels
plt.title("Normal Distribution (CLT)", fontsize=20, fontweight="bold")
plt.xlabel("Population Mean", fontsize=14, fontweight="bold")
plt.ylabel("Density", fontsize=14, fontweight="bold")

# --- Adding Grid-lines & Chart-key
plt.grid(True, linestyle="--", linewidth=0.8)
plt.legend()

# --- Shows Chart
plt.show()
```
