# **Positive Correlation**
---

- A Positive Correlation occurs when two variables move in the same direction. As the value of one variable increases, the value of the other variable tends to increase as well (and vice versa).

<br/>

#### **Key Characteristic**
- **Correlation Coefficient (r):** Ranges from 0 < r ≤ +1.0.
- **Scatter Plot Slope:** Upward sloping from left to right (↗️).
- **Relationship:** &emsp; X⬆️⟹Y⬆️ &emsp; _and_ &emsp; X⬇️⟹Y⬇️.

<br/>
<br/>

# **Negative Correlation**
---

- A Negative Correlation (also called an inverse correlation) occurs when two variables move in opposite directions. As the value of one variable increases, the value of the other variable tends to decrease.

<br/>

#### **Key Characteristics**
- **Correlation Coefficient (r):** Ranges from −1.0 ≤ r < 0.
- **Scatter Plot Slope:** Downward sloping from left to right (↘).
- **Relationship:** &emsp; X⬆️⟹Y⬇️ &emsp; _and_ &emsp; X⬇️⟹Y⬆️.
