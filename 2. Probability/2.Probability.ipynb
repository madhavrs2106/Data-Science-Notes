# **Probability**
---

- Probability measures the **likelihood of a particular outcome or event occurring**. It is typically expresed as a number 0 to 1.
- **0** indicates impossibility (event will not occur).
- **1** indicates certainty (event will occur).

# **Key Rules of Probability**
---

---
