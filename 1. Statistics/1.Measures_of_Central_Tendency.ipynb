# **Importing Neccessory Libraries**

```python
import os
import numpy as np   # --- import numpy when using numpy functions
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

file_directory = os.getcwd()
os.chdir(file_directory)
```

# **Measures of Central Tendency**

```python
reader = pd.read_csv('Titanic-Dataset.csv', chunksize=1000)
titanic_df = pd.concat(reader, ignore_index=True)
titanic_df.head()
```

## **1. Mean**

- Average value of a Column is called Mean of that Column.

- with Pandas: &emsp; &emsp; dataset['column-name'].mean()
- with NumPy: &emsp; &emsp; np.mean(dataset['column-name'])

<img src="images/Mean-Formula.png" width="400"/>

```python
meanValue = float(titanic_df['Fare'].mean())
print(f"Mean Value: {meanValue}")
```

```python
sns.histplot(x="Fare",data=titanic_df, bins=[i for i in range(0,81,10)])

# --- Mean-Line
plt.plot([meanValue for i in range(0,500)],[i for i in range(0,500)], color="red", label="Mean Line")

plt.title("Fare Distribution", fontsize=20, fontweight="bold")
plt.xlabel("Fare", fontsize=16, fontweight="bold")
plt.ylabel("Count", fontsize=16, fontweight="bold")
plt.grid(True, linestyle="--",  linewidth=0.2)
plt.legend()
plt.show()
```

## **2. Median**
- Middle value of a Column, when sorted assending is called Median of Column.

- with Pandas: &emsp; &emsp; dataset['column-name'].median()
- with NumPy: &emsp; &emsp; np.median(dataset['column-name'])

<img src="images/Median-Formula.png" width="400"/>

```python
titanic_df['Age'] = titanic_df['Age'].fillna(titanic_df['Age'].mean())
```

```python
meanValue = float(titanic_df['Fare'].mean())
```

```python
medianValue = float(titanic_df['Fare'].median())
print(f"Median Value: {medianValue}")
```

```python
sns.histplot(x='Fare', data=titanic_df, bins=[i for i in range(0,81, 10)])

# --- Median-Line
plt.plot([medianValue for i in range(0,500)], [i for i in range(0,500)],color="orange", label="Median Line")

plt.title("Fare Distribution", fontsize=20, fontweight="bold")
plt.xlabel("Fare", fontsize=16, fontweight="bold")
plt.ylabel("Count", fontsize=16, fontweight="bold")
plt.grid(True, linestyle='--', linewidth=0.2)
plt.legend()
plt.show()
```

## **3. Mode**
- Most Frequently repeated value in a Column is called Mode value of Column.

- with Pandas: &emsp; &emsp; dataset['column-name'].mode()
- with NumPy: &emsp; &emsp; np.mode(dataset['column-name'])

<img src="images/Mode-Formula.png" width="500"/>

```python
modeValue = titanic_df['Fare'].mode()[0]
print(f"Mode Value: {modeValue}")
```

```python
titanic_df['Fare'].value_counts().head(1)
```

```python
sns.histplot(x='Fare', data=titanic_df, bins=[i for i in range(0,81,10)])

# --- Mode-Line
plt.plot([modeValue for i in range(0,500)], [i for i in range(0,500)], color='green', label="Mode Line")

plt.title("Fare Distribution", fontsize=20, fontweight="bold")
plt.xlabel("Fare", fontsize=16, fontweight="bold")
plt.ylabel("Count", fontsize=16, fontweight="bold")
plt.grid(True, linestyle='--', linewidth=0.2)
plt.legend()
plt.show()
```

# **Final Graph**

This graph represents the Fare data Distributed.

- **Minimum** value for Fare is **0**.
- **Maximum** value for Fare is **80**.
- **Average** Fare is **32.20**.
- Central value of Fare is **14.45**
- Most Frequent value of Fare is **8.05**

```python
sns.histplot(x='Fare', data=titanic_df, bins=[i for i in range(0,81,10)])

# --- Mean-Line
plt.plot([meanValue for i in range(0,500)], [i for i in range(0,500)], color='red', label="Mean Line")

# --- Median-Line
plt.plot([medianValue for i in range(0,500)], [i for i in range(0,500)], color='orange', label="Median Line")

# --- Mode-Line
plt.plot([modeValue for i in range(0,500)], [i for i in range(0,500)], color='green', label="Mode Line")

plt.title("Fare Distribution", fontsize=20, fontweight="bold")
plt.xlabel("Fare", fontsize=16, fontweight="bold")
plt.ylabel("Count", fontsize=16, fontweight="bold")
plt.grid(True, linestyle='--', linewidth=0.2)
plt.legend()
plt.show()
```
