```python
# --- import Libraries to use their Functions
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# --- Changing working directory to current folder
file_directory = os.getcwd()
os.chdir(file_directory)
```

# **Percentiles & Quantiles**
---

```python
reader = pd.read_csv('Titanic-Dataset.csv', chunksize=1000)
titanic_df = pd.concat(reader, ignore_index=True)
titanic_df.head()
```

```python
# --- Cleaning
titanic_df["Age"] = titanic_df["Age"].fillna(titanic_df["Age"].mean())
titanic_df["Fare"] = titanic_df["Fare"].fillna(titanic_df["Fare"].mean())

print(f"Null Attributes in Age:   {int(titanic_df["Age"].isnull().sum())}")
print(f"Null Attributes in Fare:  {int(titanic_df["Fare"].isnull().sum())}")
```

## **Percentiles**
---

Give you a number that describes the value, that a given percent of the values are lower than.

<img src="images/Percentiles_&_Quantiles/Percentile-Formula.png" width=550/>

#### **0-th Percentile**

```python
percentile0 = float(np.percentile(titanic_df["Age"], 0))

print(
    f" 0-th Percentile:  {percentile0}\n"
    f" Minimum Value:    {titanic_df["Age"].min()}"
)
```

#### **50-th Percentile**

```python
percentile50 = float(np.percentile(titanic_df["Age"], 50))

print(
    f" 50-th Percentile:  {percentile50}\n"
    f" Median Value:      {titanic_df["Age"].median()}"
)
```

#### **100-th Percentile**

```python
percentile100 = float(np.percentile(titanic_df["Age"], 100))

print(
    f" 100-th Percentile:  {percentile100}\n"
    f" Maximum Value:      {titanic_df["Age"].max()}"
)
```

#### **Boxplot Visualization**

```python
sns.boxplot(x="Age", data=titanic_df)
plt.xlim(xmin=0, xmax=100)
plt.title("Box-Plot Visualization", fontsize=20, fontweight="bold")
plt.xlabel("Age", fontsize=14, fontweight="bold")
plt.grid(True, linestyle="--", linewidth=0.3)
plt.show()
```
