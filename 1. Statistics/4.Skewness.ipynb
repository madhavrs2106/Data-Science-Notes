# **Skewness**
---

- Skewness measures the asymmetry of the distribution.
- A skewness value of 0 indicates a perfectly symmetrical distrbution.
- **Positive skewness** indicates that the distribution is skewed to the right  (the tail is longer on the right).
- **Negative skewness** indicates that the distribution is skewed to the left  (the tail is longer on the left).

<br/>

<img src="images/Skewness/Skewness-Formula.png" width=650/>

```python
reader = pd.read_csv("Titanic-Dataset.csv", chunksize=1000)
titanic_df = pd.concat(reader, ignore_index=True)
titanic_df.head()
```

```python
# --- Cleaning
titanic_df["Age"] = titanic_df["Age"].dropna()
titanic_df["Fare"] = titanic_df["Fare"].dropna()

print(
    f" Null Attributes in Age:   {titanic_df["Age"].isnull().sum()}\n"
    f" Null Attributes in Fare:  {titanic_df["Fare"].isnull().sum()}"
)
```

## **Positive Skewness**
---

- **Positive skewness** is greater than **0**.
- **Mean** of Positive skewness lies to the **Right of both mean & median**.
- **Median** of Positive skewness is at **Middle of the data**.
- **Mode** of Positive skewness is at **Peak on the graph**.

<img src="images/Skewness/Positive-Skewness-Sequence.png" width=400/>

#### **Dataset**

```python
dataset_PS = pd.DataFrame({
    "Data" : [2,8,10,15,16,18,19,20,21,22,23,24,25,25,25,25,26,28,28,29,30,31,32,34,35,38,37,39,40,42,43,44,45,48,50,54,57,63,65,68,71,75,85,88,97]
})

dataset_PS.head()
```

#### **Positive Skewness**

```python
mean_PS = dataset_PS["Data"].mean()
median_PS = dataset_PS["Data"].median()
mode_PS = dataset_PS["Data"].mode()[0]

print(
    f" Positive Skewness:  {dataset_PS["Data"].skew()}\n\n"
    f" -->  Mean Positive-Skewness:  {mean_PS}\n"
    f" -->  Median Positive-Skewness:  {median_PS}\n"
    f" -->  Mode Skewness:  {mode_PS}"
)
```

#### **Positive Skewness Graph**

```python
# --- Setting Figure size of Chart
plt.figure(figsize = (12,6))

# --- Creating Histogram & KDE-line
sns.histplot(x="Data", data=dataset_PS, kde=True, bins=[i for i in range(0,101, 10)])

# --- Adding Mean, Median & Mode Pont-lines
plt.axvline(mean_PS, color="yellow", linestyle="-", linewidth=2, label="Mean of Positive-Skewness")
plt.axvline(median_PS, color="green", linestyle="-", linewidth=2, label="Median of Positive-Skewness")
plt.axvline(mode_PS, color="red", linestyle="-", linewidth=2, label="Mode of Positive-Skewness")

# --- Adding Title & Labels to the Graph
plt.title("Positive Skewness Graph", fontsize=20, fontweight="bold")
plt.xlabel("Values", fontsize=14, fontweight="bold")
plt.ylabel("Frequency", fontsize=14, fontweight="bold")

# --- Adding X & Y axis values (manually)
plt.xticks([i for i in range(0,101, 10)])
plt.yticks([i for i in range(0,16, 1)])

# --- Adding Grid-lines and Chart-Key
plt.grid(True, linestyle="--", linewidth=0.3)
plt.legend(loc="upper right")

# --- Shows Chart
plt.show()
```

## **Negative Skewness**
---

- **Negative Skewness** is lower than **0**.
- **Mean** of Negative Skewness lies to the **Left of both mean & median**.
- **Median** of Negative Skewness is at **Middle of the data**.
- **Mode** of Negative Skewness is at **Peak of the graph**.

<img src="images/Skewness/Negative-Skewness-Sequence.png" width=400/>

#### **Dataset**

```python
dataset_NS = pd.DataFrame({
    "Data" : [2,13,18,24,26,33,36,38,42,44,48,52,54,55,56,58,59,61,62,65,66,66,67,68,69,71,72,72,73,74,75,75,75,75,76,77,78,79,81,84,85,86,88,97,98]
})

dataset_NS.head()
```

#### **Negative Skewness**

```python
mean_NS = dataset_NS["Data"].mean()
median_NS = dataset_NS["Data"].median()
mode_NS = dataset_NS["Data"].mode()[0]

print(
    f" Negative Skewness:  {dataset_NS["Data"].skew()}\n\n"
    f" -->  Mean Negative Skewness:  {mean_NS}\n"
    f" -->  Median Negative Skewness:  {median_NS}\n"
    f" -->  Mode Skewness:  {mode_NS}"
)
```

#### **Negative Skewness Graph**

```python
# --- Setting Figure size of Chart
plt.figure(figsize = (12,6))

# --- Creating Histogram & KDE line
sns.histplot(x="Data", data=dataset_NS, kde=True, bins=[i for i in range(0,101, 10)])

# --- Adding Mean, Median & Mode Point-lines
plt.axvline(mean_NS, color="yellow", linestyle="-", linewidth=2, label="Mean of Negative-Skewness")
plt.axvline(median_NS, color="green", linestyle="-", linewidth=2, label="Median of Negative-Skewness")
plt.axvline(mode_NS, color="red", linestyle="-", linewidth=2, label="Mode of Negative-Skewness")

# --- Adding Title & Labels
plt.title("Negative Skewness Graph", fontsize=20, fontweight="bold")
plt.xlabel("Values", fontsize=14, fontweight="bold")
plt.ylabel("Frequency", fontsize=14, fontweight="bold")

# --- Adding X & Y axis values (manually)
plt.xticks([i for i in range(0,101, 10,)])
plt.yticks([i for i in range(0,16, 1)])

# --- Adding Grid-lines & Chart-Key
plt.grid(True, linestyle="--", linewidth=0.3)
plt.legend(loc="upper left")

# --- Shows Chart
plt.show()
```

## **Nornal Distribution (no-skewness)**
---

- **Normal Distribution** or **No skewness** is equal to **0**.
- **Mean**, **Median** & **Mode** of Normal Distribution lies at **Center of the data**.

<img src="images/Skewness/Normal-Distribution-Sequence.png" width=400/>

#### **Dataset**

```python
dataset_ND = pd.DataFrame({
    "Data": [4,15,18, 24,27,28,34, 35,36,38,39, 41,42,43,44,45,45,46,46,47,48,48,49, 50,50,50,50,51,52,53,54,55,56,57,57,58,59, 61,61,62,64,65, 72,74,75,76, 84,87, 98]
})

dataset_ND.head()
```

#### **Normal Distribution**

```python
mean_ND = dataset_ND["Data"].mean()
median_ND = dataset_ND["Data"].median()
mode_ND = dataset_ND["Data"].mode()[0]

print(
    f" Normal Distribution:  {dataset_ND["Data"].skew()}\n\n"
    f" -->  Mean Normal Distribution:  {mean_ND}\n"
    f" -->  Median Normal Distribution:  {median_ND}\n"
    f" -->  Mode Normal Distribution:  {mode_ND}\n"
)
```

#### **Normal Distribution Graph**

```python
# --- Setting Figure size of Chart
plt.figure(figsize = (12,6))

# --- Creating Histogram & KDE-line
sns.histplot(x="Data", data=dataset_ND, kde=True, bins=[i for i in range(0,101, 10)])

# --- Adding Mean, Median & Mode Point-lines
plt.axvline(mean_ND, color="yellow", linestyle="-", linewidth=2, label="Mean of Normal-Distribution")
plt.axvline(median_ND, color="green", linestyle="-", linewidth=2, label="Median of Normal-Distribution")
plt.axvline(median_ND, color="red", linestyle="-", linewidth=2, label="Median of Normal-Distribution")

# --- Adding Title & Labels
plt.title("Normal Distribution Graph", fontsize=20, fontweight="bold")
plt.xlabel("Values", fontsize=14, fontweight="bold")
plt.ylabel("Frequency", fontsize=14, fontweight="bold")

# --- Adding X & Y axis values (manually)
plt.xticks([i for i in range(0,101, 10)])
plt.yticks([i for i in range(0,16, 1)])

# --- Adding Grid-lines & Chart-Key
plt.grid(True, linestyle="--", linewidth=0.3)
plt.legend()

# --- Shows Chart
plt.show()
```
