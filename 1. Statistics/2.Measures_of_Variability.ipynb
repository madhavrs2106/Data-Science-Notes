# **Measures of Variability**

```python
reader = pd.read_csv('Titanic-Dataset.csv', chunksize=1000)
titanic_df = pd.concat(reader, ignore_index=True)
titanic_df.head()
```

## **1. Range**

- Difference between maximum & minimum values of dataset.
- It provides a simple measure of the spread of Data.
- It was sensitive to outliers.

<img src="images/Range-Formula.png" width="500"/>

```python
min_age = titanic_df['Age'].min()
max_age = titanic_df['Age'].max()

print(
    f" Minimum Age:  {min_age}\n"
    f" Maximum Age:  {max_age}"
)
```

```python
ageRange = max_age - min_age
print(f"Range:  {ageRange}")
```

## **2. Mean Absolute Deviation**

- Mean Absolute Deviation of dataset is the average distance between each data point and the mean.
- It gives us an idea about the variability in a dataset.
