# **Importing Neccessary Libraries**
---

```python
# --- import Libraries to use their Functions
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# --- Changing working directory to current Folder
file_directory = os.getcwd()
os.chdir(file_directory)
```

# **Measures of Variability**
---

```python
reader = pd.read_csv('Titanic-Dataset.csv', chunksize=1000)
titanic_df = pd.concat(reader, ignore_index=True)
titanic_df.head()
```

## **1. Range**
---

- Difference between maximum & minimum values of dataset.
- It provides a simple measure of the spread of Data.
- It was sensitive to outliers.

<img src="images/Measures_of_Variability/Range-Formula.png" width="550"/>

#### **Minimum**, **Maximum** & **Range**

```python
min_age = titanic_df['Age'].min()
max_age = titanic_df['Age'].max()
ageRange = max_age - min_age

print(
    f" Minimum Age:  {min_age}\n"
    f" Maximum Age:  {max_age}\n"
    f" Range:  {ageRange}"
)
```

## **2. Mean Absolute Deviation**
---

- Mean Absolute Deviation of dataset is the average distance between each data point and the mean.
- It gives us an idea about the variability in a dataset.

<img src="images/Measures_of_Variability/Mean-Absolute-Deviation.png" width="650"/>

#### **Data with same Mean**

```python
section_A = np.array([75,65,73,68,72,67])
section_B = np.array([90,47,43,96,93,51])
number = np.array([1,2,3,4,5,6])

print(
    f" Section A:  {np.mean(section_A)}\n"
    f" Section B:  {np.mean(section_B)}"
)
```

```python
# --- Creating Scatter-plot of Section-A & Section-B
plt.scatter(section_A, number, label="Section A")
plt.scatter(section_B, number, label="Section B")

# --- Adding Mean Point-line
plt.plot([np.mean(section_A) for i in range(len(number))],number, label="Mean of Sections")

# --- Adding Title & Labels
plt.title("Section Data Spread", fontsize=20, fontweight='bold')
plt.xlabel("Section Data", fontsize=14, fontweight='bold')
plt.ylabel("Number", fontsize=14, fontweight='bold')

# --- Adding Grid-lines & Chart-Key
plt.grid(True, linestyle='--', linewidth=0.8)
plt.legend()

# --- Shows Chart
plt.show()
```

#### **Absolute value**

- This converts negative value to positive, or removes negative sign from the value.

```python
print(
    f" Absolute value (10):   {np.abs(10)}\n"
    f" Absolute value (-10):  {np.abs(-10)}"
)
```

#### **Mean Absolute Deviation (MAD) Formula**

- Highest Mean Absolute Deviation (MAD) is best.

```python
mean = np.mean(section_A)

# --- Mean Absolute Deviation of Section A
mad_A = np.sum(np.abs(section_A - mean))/len(section_A)

# --- Mean Absolute Deviation of Section B
mad_B = np.sum(np.abs(section_B - mean))/len(section_B)

print(
    f" Mean Absolute Deviation of Section A:  {round(mad_A ,4)}\n"
    f" Mean Absolute Deviation of Section B:  {round(mad_B,4)}"
)
```

## **3. Standard Deviation**
---

- Standard Deviation (σ or s) is the square root of the variance. It measures how spread out the data points are relative to their mean.

<img src="images/Measures_of_Variability/Standard-Deviation-Formula.png" width=650/>

```python
std_A = np.std(section_A)
std_B = np.std(section_B)

print(
    f" Standard Deviation of Section A:  {round(std_A,4)}\n"
    f" Standard Deviation of Section B:  {round(std_B,4)}"
)
```

## **4.Variance**
---

- Variance (σ^2 or s^2) measures the average squared distance of data points from their mean. While Standard Deviation is in the original units of your data, Variance is expressed in squared units.

<img src="images/Measures_of_Variability/Variance-Formula.png" width=600/>

```python
var_A = np.var(section_A)
var_B = np.var(section_B)

print(
    f" Variance of Section A:  {round(var_A,4)}\n"
    f" Variance of Section B:  {round(var_B,4)}"
)
```

# **Age Data Spread Graph**
---

- The graph, which shows the spread of Data.

This graph represents:

- **Average** Age is **29.7**.
- Majority of data lies between **15.18** and **44.22**.

```python
avg_age = float(np.mean(titanic_df['Age']))
std_age = float(np.std(titanic_df['Age']))
var_age = float(np.var(titanic_df['Age']))

print(
    f" Average Age:  {round(avg_age,4)}\n"
    f" Standard Deviation of Age:  {round(std_age,4)}\n"
    f" Variance of Age:  {round(var_age,4)}"
)
```

```python
# --- Setting Figure size of Chart
plt.figure(figsize = (10,5))

# --- Creating Histogram
sns.histplot(x=titanic_df['Age'], data=titanic_df, bins=[i for i in range(0,81,5)])

# --- Data spreaded
plt.axvline(avg_age, color='purple', linestyle='-', linewidth=2, label=f'Average Age  ({avg_age:.2f})')
plt.axvline(avg_age - std_age, color='green', linestyle=':', linewidth=2, label=f'-1 Std Age  ({avg_age-std_age:.2f})')
plt.axvline(avg_age + std_age, color='red', linestyle=':', linewidth=2, label=f'+1 Std Age  ({avg_age+std_age:.2f})')

# --- Adding Title & Labels
plt.title("Age Data Spread", fontsize=20, fontweight='bold')
plt.xlabel("Age", fontsize=14, fontweight='bold')
plt.ylabel("Number", fontsize=14, fontweight='bold')

# --- Adding Grid-lines & Chart-Key
plt.grid(True, linestyle='--', linewidth=0.2)
plt.legend()

# --- Shows Chart
plt.show()
```
