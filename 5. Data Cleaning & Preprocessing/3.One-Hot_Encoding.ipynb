# **One-Hot Encoding**
---
<br/>

### **Definition**
---

- One-hot encoding is a simple way to convert text or category labels into numbers (**0s** and **1s**) so that computer and machine learning models can understand them

<br/>

### **Importting Neccessary Libraries**
---

```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder
```

### **Importting Dataset**
---

```python
loan_df = pd.read_csv("loan_data_set.csv")

loan_df.head()
```

### **Exploring Data**
---

- To perform One-Hot Encoding, we have to find Categorical attributes (or Columns) like here 'Gender' & 'Married' are Categorical attributes.

```python
loan_df.isnull().sum()
```

### **Filling Null-Values**
---

```python
# --- filling null-values in 'Gender' with mode of 'Gender'
loan_df["Gender"] = loan_df["Gender"].fillna(loan_df["Gender"].mode()[0])

loan_df["Gender"].isnull().sum()
```

```python
# --- filling null-values in 'Married' with mode of 'Married'
loan_df["Married"] = loan_df["Married"].fillna(loan_df["Married"].mode()[0])

loan_df["Married"].isnull().sum()
```

### **One-Hot Encoding**
----

```python
encoding_data = loan_df[["Gender","Married"]]

encoding_data.head()
```

- #### **Using Pandas (_get_dummies_)**

```python
pd.get_dummies(encoding_data)
```

```python
pd.get_dummies(encoding_data).info()
```

- #### **Using Scikit-Learn**

```python
ohe = OneHotEncoder()
memory_address = ohe.fit_transform(encoding_data)
memory_address
```

```python
arr = memory_address.toarray()
arr
```

```python
pd.DataFrame(arr, columns=["Gender_Female","Gender_Male","Married_Female","Married_Male"])
```
