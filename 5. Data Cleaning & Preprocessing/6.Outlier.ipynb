# **Outlier**
---
<br/>

### **Definition**
---

- **Outlier** is a data point, person, or thing that is very different from the rest of its group. It lies far outside the normal pattern.

### **Importting Neccessary Libraries**
---

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

### **Importting Dataset**

```python
loan_df = pd.read_csv("loan_data_set.csv")

loan_df.head()
```

##### **Working on 'ApplicantIncome' (already cleaned)**
<br/>

here we saw:
-  mean of 'ApplicantIncome' is **5403.45**
-  min of 'ApplicantIncome' is **150**

but,
-  max of 'ApplicantIncome' is **81000**

<br/>

difference between:
-  mean & min &emsp; : &emsp; **5253.45**
-  mean & max &emsp; : &emsp; **75596.55**

```python
applicantIncome_min = loan_numtype_df["ApplicantIncome"].min()
applicantIncome_max = loan_numtype_df["ApplicantIncome"].max()
applicantIncome_mean = loan_numtype_df["ApplicantIncome"].mean()

print(
    f" ->   ApplicantIncome min  -  {applicantIncome_min}\n"
    f" ->   ApplicantIncome max  -  {applicantIncome_max}\n"
    f" ->   ApplicantIncome mean -  {applicantIncome_mean}"
)
```

### **Outlier**
---

##### **From Boxplot**

```python
plt.figure(figsize=(16,6))
sns.boxplot(x="ApplicantIncome", data=loan_numtype_df)
plt.show()
```

##### **From Distribution-chart**

```python
plt.figure(figsize=(15,6))
sns.histplot(x="ApplicantIncome", data=loan_numtype_df, kde=True)
plt.show()
```
